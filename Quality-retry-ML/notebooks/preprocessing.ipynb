```python
import numpy as np
import pandas as pd
import warnings
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("..\\data\\raw\\df.csv")
df.head(5)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(round(df.corr(numeric_only=True),2), annot=True, cmap="coolwarm", linewidths=0.7)
plt.title("Correlation Heatmap")


plt.show()
```

```python
df.drop('Smart_Work_Index',axis=1,inplace=True)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(round(df.corr(numeric_only=True),2), annot=True, cmap="coolwarm", linewidths=0.7)
plt.title("Correlation Heatmap")


plt.show()
```

```python
df['Prod_Efficiency'] = df['Productivity_Score'] * df['Efficiency_Rating']
df['Exp_Task_Rate'] = df['Years_Experience'] * df['Task_Completion_Rate']
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(round(df.corr(numeric_only=True),2), annot=True, cmap="coolwarm", linewidths=0.7)
plt.title("Correlation Heatmap")


plt.show()
```

```python
df.drop(columns=['Productivity_Score', 'Years_Experience'], inplace=True,axis=1)
```

```python
plt.figure(figsize=(4, 8))
sns.heatmap(df.corr(numeric_only=True)[['Quality_Score']], annot=True, cmap="coolwarm")

plt.show()
```
