```python
fig ,ax = plt.subplots(1,2,figsize=(10,4))

lavozim_tartibi = ['Junior', 'Mid-Level', 'Senior', 'Lead', 'Manager', 'Director']

sns.barplot(data=df,x='Job_Level',y='Productivity_Score',errorbar=None, ax=ax[0],order=lavozim_tartibi,hue='Job_Level',palette='Set2')
sns.barplot(data=df,x='Job_Level',y='Quality_Score',errorbar=None, ax=ax[1],order=lavozim_tartibi,hue='Job_Level',palette='Set2')
ax[0].set_ylim(75,95)
ax[1].set_ylim(75,95)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cat_cols = df.select_dtypes(include=['object', 'category']).columns

# Har bir ustun uchun alohida grafik chizamiz
for col in cat_cols:
    plt.figure(figsize=(6, 3))
    
    # Ustunlar soni ko'p bo'lsa (masalan, Department), y='col' qilib gorizontal chizgan ma'qul
    sns.countplot(data=df, y=col, order=df[col].value_counts().index, palette='viridis',hue=col,legend=False)
    
    plt.title(f"{col} ustunining taqsimoti")
    plt.xlabel("Xodimlar soni")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```
